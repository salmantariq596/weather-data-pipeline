# Weather Data Analysis - Bahawalpur

This notebook connects to the PostgreSQL database and performs exploratory data analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine

# Connect to database
engine = create_engine('postgresql://weather_user:weather_pass@postgres:5432/weather_db')

# Load data
df = pd.read_sql('SELECT * FROM stg_weather_current ORDER BY fetched_at DESC LIMIT 1000', engine)
df.head()
```

```python
# Temperature trends
plt.figure(figsize=(12, 6))
plt.plot(df['fetched_at'], df['temperature'], label='Temperature')
plt.plot(df['fetched_at'], df['feels_like'], label='Feels Like')
plt.xlabel('Time')
plt.ylabel('Temperature (°C)')
plt.title('Temperature Trends - Bahawalpur')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap
corr_cols = ['temperature', 'humidity_percent', 'pressure_hpa', 'wind_speed_ms']
corr = df[corr_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0)
plt.title('Weather Metrics Correlation')
plt.show()
```
